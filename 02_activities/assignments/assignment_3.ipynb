#### **Question 2:** 
#### Data-visualization

Let's create plots to visualize the relationships between the features (e.g., alcohol, malic acid, ash...).


```python
def plot_feature_pairs(data, feature_names, color_labels=None, title_prefix=''):
    """
    Helper function to create scatter plots for all possible pairs of features.
    
    Parameters:
    - data: DataFrame containing the features to be plotted.
    - feature_names: List of feature names to be used in plotting.
    - color_labels: Optional. Cluster or class labels to color the scatter plots.
    - title_prefix: Optional. Prefix for plot titles to distinguish between different sets of plots.
    """
    # Create a figure for the scatter plots
    plt.figure(figsize=(60, 60))
    
    # Counter for subplot index
    plot_number = 1
    
    # Loop through each pair of features
    for i in range(len(feature_names)):
        for j in range(i + 1, len(feature_names)):
            plt.subplot(len(feature_names)-1, len(feature_names)-1, plot_number)
            
            # Scatter plot colored by labels if provided
            if color_labels is not None:
                plt.scatter(data[feature_names[i]], data[feature_names[j]], 
                            c=color_labels, cmap='viridis', alpha=0.7)
            else:
                plt.scatter(data[feature_names[i]], data[feature_names[j]], alpha=0.7)
            
            plt.xlabel(feature_names[i])
            plt.ylabel(feature_names[j])
            plt.title(f'{title_prefix}{feature_names[i]} vs {feature_names[j]}')
            
            # Increment the plot number
            plot_number += 1

    # Adjust layout to prevent overlap
    plt.tight_layout()

    # Show the plot
    plt.show()

# Get feature names
feature_names = wine_df.columns

# Use the helper function to plot scatter plots without coloring by cluster labels
plot_feature_pairs(wine_df, feature_names, title_prefix='Original Data: ')
```

#### **Question 3:** 
#### Data cleaning

```python
# Initialize the StandardScaler
scaler = StandardScaler()

# Scale all the features in the dataset
scaled_features = scaler.fit_transform(wine_df)

# Create a new DataFrame with scaled features
scaled_wine_df = pd.DataFrame(scaled_features, columns=wine_data.feature_names)

# Display the first few rows of the scaled DataFrame
print(scaled_wine_df.head())
```

Why is it important to standardize the features of a dataset before applying clustering algorithms like K-Means? Discuss the implications of using unstandardized data in your analysis.

**ANSWER:**   
It is important to standardize the features of a dataset before applying clustering algorithms in order to:
- ensure that all features contribute equally to the distance calculations and therefore grouping/clustering process;   
- prevent any single feature/variable from dominating the results due to its scale;   
- ensure fair contribution from all variables.   

The implications of using unstandardized data includes the risk that one variable could dominate the analysis and visual plot(s), making it difficult to identify distinct clusters/groupings of datapoints, particularly when feature ranges differ substantially.

#### **Question 4:** 
#### K-means clustering 

Apply the K-Means clustering algorithm to the Wine dataset. Choose the value 3 for the number of clusters (`k=3`) and fit the model. Assign cluster labels to the original data and add them as a new column in the DataFrame.

```python
#ANSWER:
kmeans = KMeans(n_clusters=3, random_state=0, n_init=10)

wine_clusters = kmeans.fit(scaled_features)

#Assigning cluster labels
wine_clusters.labels_
```

```python
scaled_wine_with_clusters = scaled_wine_df.copy()

scaled_wine_with_clusters['wine_clusters'] = wine_clusters.labels_

scaled_wine_with_clusters
```

```python
def plot_feature_pairs(data, feature_names, color_labels=None, title_prefix=''):
    """
    Helper function to create scatter plots for all possible pairs of features.
    Parameters:
    - data: DataFrame containing the features to be plotted.
    - feature_names: List of feature names to be used in plotting.
    - color_labels: Optional. Cluster or class labels to color the scatter plots.
    - title_prefix: Optional. Prefix for plot titles to distinguish between different sets of plots.
    """
    # Create a figure for the scatter plots
    plt.figure(figsize=(60, 60))
    
    # Counter for subplot index
    plot_number = 1
    
    # Loop through each pair of features
    for i in range(len(feature_names)):
        for j in range(i + 1, len(feature_names)):
            plt.subplot(len(feature_names)-1, len(feature_names)-1, plot_number)
            
            # Scatter plot colored by labels if provided
            if color_labels is not None:
                plt.scatter(data[feature_names[i]], data[feature_names[j]], 
                            c=color_labels, cmap='viridis', alpha=0.7)
            else:
                plt.scatter(data[feature_names[i]], data[feature_names[j]], alpha=0.7)
            
            plt.xlabel(feature_names[i])
            plt.ylabel(feature_names[j])
            plt.title(f'{title_prefix}{feature_names[i]} vs {feature_names[j]}')
            
            # Increment the plot number
            plot_number += 1

    # Adjust layout to prevent overlap
    plt.tight_layout()

    # Show the plot
    plt.show()

# Get feature names
feature_names = scaled_wine_with_clusters.columns

# Use the helper function to plot scatter plots, colored by cluster labels
plot_feature_pairs(scaled_wine_with_clusters, feature_names, color_labels=scaled_wine_with_clusters['wine_clusters'], title_prefix='Clustered Data: ')
```

We chose `k=3` for the number of clusters arbitrarily. However, in a real-world scenario, it is important to determine the optimal number of clusters using appropriate methods.

**Question**: What is one method commonly used to determine the optimal number of clusters in K-means clustering, and why is this method helpful?

**ANSWER:**  
We can find the optimal number of clusters (K) by calculating the WSSD (Within-Cluster Sum of Squared Distances, aka: inertia) formula, and then using the elbow method plot. The elbow method plots the total WSSD against different values of K, and then looking for an “elbow” in the plot, which is the point where the plot bends and flattens out (like the tangent of a curve), showing that increasing the number of clusters no longer leads to a significant reduction in WSSD.
This method is helpful because it measures cluster quality, leading to better visualization of data groupings. 

#### **Question 5:** 
#### Bootstrapping 

Implement bootstrapping on the mean of `color intensity`. Generate 10000 bootstrap samples, calculate the mean for each sample, and compute a 90% confidence interval.

```python

np.random.seed(101)
color_intensity_sample = wine_df.sample(n=40)
#color_intensity_sample

color_intensity_sample['color_intensity'].mean()

bootstrap_wine = []

for i in range(10000):
    sample = color_intensity_sample.sample(frac=1, replace=True)  
    sample = sample.assign(replicate=i) 
    bootstrap_wine.append(sample) 

boot10000 = pd.concat(bootstrap_wine)
#boot10000

mean_color_intensity = boot10000.groupby("replicate")['color_intensity'].mean().reset_index(name = 'mean_color_intensity')
#mean_color_intensity

lower_bound = mean_color_intensity["mean_color_intensity"].quantile([0.1])
upper_bound = mean_color_intensity["mean_color_intensity"].quantile([0.9])

# Display the result
print(f"Mean of Color Intensity: {mean_color_intensity}")
print(f"90% Confidence Interval of Mean Color Intensity: ({lower_bound}, {upper_bound})")
```

**Question:**
- Why do we use bootstrapping in this context? What does it help us understand about the mean?

**ANSWER:**   
We use bootstrapping from the data we already have to estimate the mean of the colour intensity of the entire population of wine, as we only have a sample (not the complete population of all wines) and we want to assess how closely the sample mean approximates the true population mean of wine colour intensity, by calculating a distribution of possible means. This distribution helps us understand how much our estimate might vary and gives us a range of likely values.

**Question:**
- What is the purpose of calculating the confidence interval from the bootstrap samples? How does it help us interpret the variability of the estimate?

**ANSWER:**   
We calculate the confidence interval from the bootstrap samples in order to estimate the variability and precision of wine colour intensity without needing to have the data from the entire population of wine. 
Confidence intervals help us interpret variability by giving us an idea of how likely our sample is to resemble the population. 

**Question:**

- Reflect on the variability observed in the bootstrapped means and discuss whether the mean of the color intensity appears to be a stable and reliable estimate based on the confidence interval and the spread of the bootstrapped means.

**ANSWER:**   
The 90% confidence interval calculated above suggests that the true population mean of wine colour intensity has a 90% likelihood of being between 4.5455 and 5.49855. The calculated mean of colour intensity of all wine from the original wine_df dataset (5.0581) does in fact fall in the middle of this range (see cell below). 
Looking at the histogram of colour intensity means generated by the bootstrap samples (above), the histogram shows a bell curved distribution, which suggests a reasonable spread of the bootstrapped means.
