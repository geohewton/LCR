### Classification using KNN

Let's set up our workspace and use the **Wine dataset** from `scikit-learn`. This dataset contains 178 wine samples with 13 chemical features, used to classify wines into different classes based on their origin.

The **response variable** is `class`, which indicates the type of wine. We'll use all of the chemical features to predict this response variable.

```python
# Import standard libraries
import pandas as pd
import numpy as np
import random
import matplotlib.pyplot as plt
import matplotlib.colors as mcolors
from sklearn.preprocessing import StandardScaler
from sklearn.model_selection import train_test_split
from sklearn.neighbors import KNeighborsClassifier
from sklearn.metrics import recall_score, precision_score
from sklearn.model_selection import cross_validate
from sklearn.model_selection import GridSearchCV
from sklearn.metrics import accuracy_score
```

```python
from sklearn.datasets import load_wine

# Load the Wine dataset
wine_data = load_wine()

# Convert to DataFrame
wine_df = pd.DataFrame(wine_data.data, columns=wine_data.feature_names)

# Bind the 'class' (wine target) to the DataFrame
wine_df['class'] = wine_data.target

# Display the DataFrame
wine_df
```

#### **Question 2:** 
#### Standardization and data-splitting

Next, we must preform 'pre-processing' or 'data munging', to prepare our data for classification/prediction. For KNN, there are three essential steps. A first essential step is to 'standardize' the predictor variables. We can achieve this using the scaler method, provided as follows:

```python
# Select predictors (excluding the last column)
predictors = wine_df.iloc[:, :-1]

# Standardize the predictors
scaler = StandardScaler()
predictors_standardized = pd.DataFrame(scaler.fit_transform(predictors), columns=predictors.columns)

# Display the head of the standardized predictors
print(predictors_standardized.head())
```

```python
predictors_standardized
```

```python
#Insert response variable column class back into dataframe (reference https://pandas.pydata.org/docs/reference/api/pandas.DataFrame.assign.html)

predictors_standardized.insert(13,'class', wine_df['class'])
```

```python
predictors_standardized
```

(i) Why is it important to standardize the predictor variables?

ANSWER:
It is important to standardize predictor variables in order to:
- ensure fair contribution from all predictor variables;
- allow the model to consider each variable equally;
- reduce the effect of class imbalance between variables and ensure that no single variable disproportionately affects the model due to its scale/weight. 

(ii) Why did we elect not to standard our response variable `Class`?

ANSWER: We did not standardize the response variable 'class' because the values of 'class' are integer categories that are dependent on the predictor variables. 'Class' is the variable of interest that is being predicted by the other variables.

(iii) A second essential step is to set a random seed. Do so below (Hint: use the random.seed function). Why is setting a seed important? Is the particular seed value important? Why or why not?

```python
np.random.seed(42)
```

ANSWER: Random numbers are used to test code in order to remove human bias from data selection. Setting a random seed is done to generate the same random numbers for use in testing your code each time you run it (ensuring reproducibility and consistency), basically to replicate our findings. Setting a seed guarantees that you generate the exact same sequence of random numbers so that you get the same results every time the code is executed.
The particular seed value used is not important (and any integer value can be used) because each seed value has its own sequence of random numbers. As long as the random number generator is later reinitialized with the same seed value, it will produce the same sequence of numbers so that anyone running that code will get the same results.

(iv) A third essential step is to split our standardized data into separate training and testing sets. We will split into 75% training and 25% testing. The provided code randomly partitions our data, and creates linked training sets for the predictors and response variables. 

Extend the code to create a non-overlapping test set for the predictors and response variables.

```python
# set a seed for reproducibility
np.random.seed(42)

# split the data into a training and testing set. hint: use train_test_split !
wine_df_train, wine_df_test = train_test_split(
    predictors_standardized, train_size=0.75, shuffle=True, stratify=predictors_standardized["class"]
)

# Your code here ...
wine_df_train.info()

#wine_df_test.info() in next cell.
```

```python
wine_df_test.info()
```

#### **Question 3:**
#### Model initialization and cross-validation
We are finally set to fit the KNN model. 


Perform a grid search to tune the `n_neighbors` hyperparameter using 10-fold cross-validation. Follow these steps:

1. Initialize the KNN classifier using `KNeighborsClassifier()`.
2. Define a parameter grid for `n_neighbors` ranging from 1 to 50.
3. Implement a grid search using `GridSearchCV` with 10-fold cross-validation to find the optimal number of neighbors.
4. After fitting the model on the training data, identify and return the best value for `n_neighbors` based on the grid search results.

```python
# Step 1
knn = KNeighborsClassifier(n_neighbors=10)
```

```python
X = wine_df_train[["total_phenols","flavanoids"]]
y = wine_df_train["class"]
```

```python
knn.fit(X,y)
```

```python
wine_df_test["predicted"] = knn.predict(wine_df_test[["total_phenols","flavanoids"]])
```

```python
wine_df_test[["predicted","class"]]
```

```python
pd.crosstab(
    wine_df_test["class"],
    wine_df_test["predicted"]
)
```

```python
knn = KNeighborsClassifier(n_neighbors=10)
X = wine_df_train[["total_phenols","flavanoids"]]
y = wine_df_train["class"]
```

```python
returned_dictionary = cross_validate(
    estimator= knn,
    cv= 10,
    X=X,
    y=y
)
```

```python
cv_10_df = pd.DataFrame(returned_dictionary)
cv_10_df
```

```python
cv_10_df.agg(["mean","sem"])
```

```python
#Step 2: Define a parameter grid for n_neighbors ranging from 1 to 50.

parameter_grid = {
    "n_neighbors": range(1,50,1)
}
```

```python
#Step 3: Implement grid search using GridSearchCV with 10-fold cross-validation to find the optimal number of neighbors.

wine_tune_grid = GridSearchCV(
    estimator= knn,
    param_grid=parameter_grid,
    cv=10
)
```

```python
wine_tune_grid.fit(
    wine_df_train[["total_phenols","flavanoids"]],
    wine_df_train["class"]
)
```

```python
accuracy_grid = pd.DataFrame(wine_tune_grid.cv_results_)
accuracy_grid
```

```python
#Create plot
plt.figure(figsize=(10,6))

plt.plot(accuracy_grid['param_n_neighbors'], accuracy_grid['mean_test_score'], '-o', color='blue')

plt.xlabel('Number of neighbours')
plt.ylabel('Accuracy Est.')
plt.title('KNN Performance')
plt.tight_layout()
plt.show()
```

```python
wine_tune_grid.best_params_
```

#### **Question 4:**
#### Model evaluation

Using the best value for `n_neighbors`, fit a KNN model on the training data and evaluate its performance on the test set using `accuracy_score`.

```python
knn = KNeighborsClassifier(n_neighbors=25)
```

```python
X = wine_df_train[["total_phenols","flavanoids"]]
y = wine_df_train["class"]
```

```python
knn.fit(X,y)
```

```python
wine_df_test["predicted"] = knn.predict(wine_df_test[["total_phenols","flavanoids"]])
```

```python
wine_df_test[["predicted","class"]]
```

```python
pd.crosstab(
    wine_df_test["class"],
    wine_df_test["predicted"]
)
```

```python
accuracy_score(
    y_true=wine_df_test["class"], 
    y_pred=wine_df_test["predicted"]
    )
```

ANSWER: This model using the predictor variables "total_phenols" and "flavanoids" and the 25 nearest neighbours produced an accuracy of 86.7% for the prediction of wine class.
